**try-except-finally** 

```python
try:
    print("try block")
    num1=eval(input("Enter num1: "))
    num2=eval(input("Enter num1: "))
    add=num1+num2
    print(add)

except Exception as e:
    print("Exception block")
    print(e)
    a=100
    b=200
    print(a+b)
    
finally:
    print("Finally block")
```

```python
try:
    print("try block")
    num1=eval(input("Enter num1: "))
    num2=eval(input("Enter num1: "))
    add=num1+num2
    print(add)

except Exception as e:
    print("Exception block")
    print(e)
    a=100
    b=200
    print(a+b)
    

print("Finally block")
```

```python
# wap ask the user enter a number
# print if it is even number or odd number
# try-except
try:
    num1=eval(input("enter a number:"))
    if num1%2==0:
        print(f"{num1} is even")
    else:
        print(f"{num1} is an odd number")
except Exception as e:
    print(e)
```
