$Method-4$

**Apply eval at the time addition**

```python
num1=input("enter the number1:") # '100.5'
num2=input("enter the number2:") # '100'
#eval(num1+num2)  # eval('100.5'+'100')= eval('100.5100')=100.51
add=eval(num1)+eval(num2)  #  eval('100.5')+eval('100'):100.5+100
print("The addition of {} and {} is {}".format(num1,num2,add))
print(f"The addition of {num1} and {num2} is {add}")
print(f"The addition of {num1} and {num2} is {eval(num1)+eval(num2)}")
```

```python
bill=eval(input("enter the bill:"))
tip=eval(input("enter the tip in %:"))
tip_amount= (tip*bill)/100
total_bill=bill+tip_amount
print("the total bill is:",total_bill)
```
