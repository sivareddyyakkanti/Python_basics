```python
# if you want print first 10 numbers in for loop
for i in range(1,11):
    print(i,end=' ')


i=1  # start
while <condition>:
    print(i,end=' ')
    i=i+1  # increment

# You need to write a condition
# if that condition is true then only
# will enter inside the while loop

# i<11  i<10 i<9
# i<=10
# True
# i!=0
```

```python
i=1  # start
while i<11:
    print(i,end=' ')
    i=i+1  # increment

# step-1:  i=1   while 1<10 (T)   print(1)   i=1+1=2
# step-2:  i=2   while 2<10 (T)  print(2)    i=2+1=3
#
#step-9:   i=9  while 9<10  (T)  print(9)  i=9+1=10
# step-10  i=10  while 10<10 F
```

```python
i=1
while i<11:
    print(i,end=' ')
    i=i+1
```

```python
s=1
while s==10:
    print(s,end=' ')
    s=s+1

# step-1: s=1  while 1==10 F
```

```python
########################### 1 to 10 ##############################
i=1  # start
while i<11:
    print(i,end=' ')
    i=i+1 

########################   -1 to  -10###############################
i=-1
while i>-11:
    print(i,end=' ')
    i=i-1

###################### -10 to -1 ##################################
i=-10
while i<0:
    print(i,end=' ')
    i=i+1

######################## True keyword 1to10##############################
i=1
while True:
    print(i,end=' ')
    i=i+1
    if i==10:
        break
```

```python
# 2Q)  WAP take the 5 random numbers between 20 to 50 in a loop and print the sqaure of the number
import random
i=1
while i<=5:
    n=random.randint(10,20)
    print(f"Square of {n} is {n*n}")
    i=i+1
```
