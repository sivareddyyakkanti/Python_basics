```python
gender=input("Enter Gender")
if(gender == 'male'):
    age=eval(input("Enter Age:"))
    if(age>30):
        print("Man")
    else:
        print("Boy")
elif(gender == "female"):
    age=eval(input("Enter Age:"))
    if(age>30):
        print("Woman")
    else:
        print("Girl")
else:
    print("Enter valid information")
```
