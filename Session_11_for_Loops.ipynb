$Method-3$

**range(start,stop,step)**

- start= what we mentioned 

- step = increment or decrement

    - if step number (+)ve increment

        - last= stop-1

    - if step number (-)ve  decrement

        - last= stop+1

```python
# case-1
for i in range(2,20,2):
    print(i,end=' ')

# start=2
# step=2  +ve
# last=stop-1=20-1=19
# start with 2 , goes till 19  with gap =2 
# 2  4  6  8  10  12  14  16  18  
```

```python
for i in range(2,20,3):
    print(i,end=' ')

# start=2
# step=3  +ve
# last=20-1=19
# 2  5  8  11  14  17 
```

```python
# case-2
for i in range(2,20,-2):
    print(i,end=' ')

# start=2
# step= -2  neg
#  last = 20+1=21

# No error - no answer
#      -1 0  1 2
```

```python
for i in range(20,2,-2):
    print(i,end=' ')

# start=20
# step= -2   negative
# last= 2+1=3

# 20  18  16  14  12  10  8  6  4 
```
