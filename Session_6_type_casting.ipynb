## Type caste

**Data type conversion**

converting one data type to another data type

**Integer type conversion**

- int to float

- int to str

- int to bool

**Float type conversion**
- float to int

- float to str

- float to bool

**string type conversion**
- str to int

- str to float

- str to bool


**bool type conversion**
- bool to int

- bool to float

- bool to str

```python
name=100
print("float convesrion is:",float(name))
print("str convesrion is:",str(name))
print("bool convesrion is:",bool(name))
```

```python
float(name),str(name), bool(name)
```

```python
num=-100
float(num),str(num), bool(num)
# -100.0, '-100', True
```

```python
num=0
float(num),str(num), bool(num)
# 0.0, '0', False
```

**Note**

- For Integer 0 value we will get False

- Remaining all other values (+)ve  and (-)ve wil get True only
